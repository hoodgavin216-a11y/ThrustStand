## **Question - Which product category and location should we target for our next ad campaign?**

```python
# Import required python packages
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Run this cell to import the sales dataset
data_url = 'https://raw.githubusercontent.com/datastray/python_project/main/sales_data.csv'
df = pd.read_csv(data_url)

# View first 5 rows of the dataset
df.head()
```

```python
# Inspect your data
df.info()
```

```python
# Get product category revenue totals
product_category_revenue = df.groupby('product_category')['total_revenue'].sum()

#get product category revenue and proportions
total_revenue = df['total_revenue'].sum()
product_category_percentage = (product_category_revenue/total_revenue) * 100

#print
product_category_percentage
```

```python
# View product category splits in a chart
product_category_percentage.plot(kind='bar')
plt.title('percentage of total revaune by product category')
plt.ylabel('precentage (%)')
plt.show()
```

```python
# what about a combination of location and product category?
#group by both product category and location
product_location_revenue=df.groupby(['product_category', 'location'])['total_revenue'].sum()
product_location_revenue
```

```python
product_location_revenue.plot(kind='barh')
plt.title('gavins experiement')
plt.ylabel('revenue')
plt.xlabel('names')
plt.show()
```
